```python
from dash import Dash, dcc, html, Input, Output, dash_table
import plotly.express as px
import pandas as pd
import requests

# Initialize Dash app
app = Dash(__name__)

# Columns expected from the API
columns = [
    "id", "name", "breed", "age_upon_outcome_weeks", "sex_upon_outcome",
    "animal_type", "color", "date_of_birth", "datetime", "monthyear",
    "outcome_type", "outcome_subtype", "location_lat", "location_long"
]

# Layout
app.layout = html.Div([
    html.Img(
        src='/assets/Grazioso Salvare Logo.png',
        style={'height': '150px', 'width': '300px', 'margin': 'auto', 'display': 'block'}
    ),
    html.Center(html.B(html.H1('Grazioso Salvare Rescue Dashboard - Ty Wheeler'))),

    dcc.RadioItems(
        id='rescue-filter',
        options=[
            {'label': 'Water Rescue', 'value': 'water'},
            {'label': 'Mountain Rescue', 'value': 'mountain'},
            {'label': 'Disaster Rescue', 'value': 'disaster'},
            {'label': 'Reset Filter', 'value': 'reset'}
        ],
        value='reset',
        style={'padding': '20px'}
    ),

    dash_table.DataTable(
        id='datatable-id',
        columns=[{"name": col, "id": col} for col in columns],
        page_action="native",
        page_current=0,
        page_size=10,
        style_table={'overflowX': 'auto'}
    ),

    html.Br(),
    html.Hr(),

    html.Div([
        dcc.Graph(id='geo-chart', className='col s6'),
        dcc.Graph(id='outcome-chart', className='col s6')
    ], style={'display': 'flex'})
])

# Callback using authenticated session to API
@app.callback(
    [Output('datatable-id', 'data'),
     Output('geo-chart', 'figure'),
     Output('outcome-chart', 'figure')],
    [Input('rescue-filter', 'value')]
)
def update_dashboard(rescue_type):
    try:
        # Start the session and log in
        session = requests.Session()
        login_response = session.post("http://127.0.0.1:5000/login", json={
            "username": "admin",
            "password": "adminpass"
        })

        if not login_response.ok:
            print("Login failed:", login_response.text)
            return [], px.scatter_mapbox().update_layout(title="Login Failed"), px.bar().update_layout(title="Login Failed")

        # Retrieve data using authenticated session
        response = session.get("http://127.0.0.1:5000/dogs")
        data = response.json()
        df = pd.DataFrame(data)

        # Rescue type filtering logic
        if rescue_type != 'reset':
            rescue_criteria = {
                'water': {
                    'breeds': ["Labrador Retriever Mix", "Chesapeake Bay Retriever", "Newfoundland"],
                    'max_age': 104
                },
                'mountain': {
                    'breeds': ["German Shepherd", "Alaskan Malamute", "Old English Sheepdog", "Rottweiler", "Siberian Husky"],
                    'max_age': 156
                },
                'disaster': {
                    'breeds': ["Doberman Pinscher", "German Shepherd", "Golden Retriever", "Bloodhound", "Rottweiler"],
                    'max_age': 208
                }
            }
            criteria = rescue_criteria[rescue_type]
            df = df[df['breed'].isin(criteria['breeds']) & (df['age_upon_outcome_weeks'] <= criteria['max_age'])]

        # Drop bad rows
        df = df.dropna(subset=["location_lat", "location_long", "outcome_type"])

        if df.empty:
            return [], px.scatter_mapbox().update_layout(title="No Data"), px.bar().update_layout(title="No Outcomes")

        # Map
        geo_fig = px.scatter_mapbox(
            df,
            lat='location_lat',
            lon='location_long',
            hover_name='breed',
            hover_data=['name', 'outcome_type'],
            zoom=9,
            height=400
        )
        geo_fig.update_layout(mapbox_style="open-street-map")

        # Outcome chart
        outcome_counts = df['outcome_type'].value_counts().reset_index(name='count')
        outcome_counts.rename(columns={'index': 'outcome_type'}, inplace=True)
        outcome_fig = px.bar(
            outcome_counts,
            x='outcome_type',
            y='count',
            labels={'outcome_type': 'Outcome Type', 'count': 'Count'},
            title='Outcome Types'
        )

        return df.to_dict("records"), geo_fig, outcome_fig

    except Exception as e:
        print(f"Dashboard error: {e}")
        return [], px.scatter_mapbox().update_layout(title="Error"), px.bar().update_layout(title="Error")

# Launch the app
import webbrowser
webbrowser.open("http://127.0.0.1:8050")
app.run(debug=True)
```
